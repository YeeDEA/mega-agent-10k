# 10K Records Analysis Notebook

This notebook loads the `population_10k.csv` and `traffic_10k.csv` datasets, provides summary statistics, creates visualizations using **matplotlib** and **plotly**, and generates a simple interactive dashboard saved as `dashboard.html`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.io as pio

# Load datasets (CSV files should be placed in the same directory as this notebook)
pop_path = 'population_10k.csv'
traffic_path = 'traffic_10k.csv'

df_pop = pd.read_csv(pop_path)
df_traffic = pd.read_csv(traffic_path)

# Quick preview
print('Population data shape:', df_pop.shape)
print('Traffic data shape:', df_traffic.shape)
```

```python
# Summary statistics
print('Population dataset summary:')
print(df_pop.describe())

print('\nTraffic dataset summary:')
print(df_traffic.describe())
```

```python
# Matplotlib histogram of population
plt.figure(figsize=(8,5))
plt.hist(df_pop['population'], bins=30, color='skyblue', edgecolor='black')
plt.title('Population Distribution (Matplotlib)')
plt.xlabel('Population')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

# Plotly scatter for traffic
fig = px.scatter(df_traffic, x='hour', y='traffic_volume', title='Traffic Volume by Hour')
fig.show()
```
